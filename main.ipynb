## 3. Análise descritiva

### 3.1 Quantos jogadores únicos existem na base?

```python
qtd_jogadores_unicos = plan['player'].nunique()
jogadores_unicos = plan['player'].unique()
print(qtd_jogadores_unicos)
jogadores_unicos
```

## 3.2 Média de pontos, reb e ast's por player

```python
pts = plan.groupby('player')['pts'].mean()
pts
```

```python
ast = plan.groupby('player')['ast'].mean()
ast
```

```python
trb = plan.groupby('player')['trb'].mean()
trb
```

### 3.3 Quem é o jogador com mais pontos totais na temporada

```python
temporada = plan[plan['season'] == 2025]
mais_pontos = temporada.loc[temporada['pts'].idxmax()]
mais_pontos
```

```python
mais_pontos[['player', 'pts']]
```

### 3.4 Descobrindo o jogador com o maior PPG

```python
plan['ppg'] = plan['pts'] / plan['g']
```

```python
ppg = plan.loc[plan['ppg'].idxmax()]
ppg
```

## 4. Ranking

### 4.1 Top 10 jogadores com mais ast na temporada

```python
ast_df_temp = temporada.groupby('player')['ast'].sum()
ast_df_temp = pd.DataFrame(ast_df_temp)
ast_df_temp
```

```python
# ast_df_temp.nlargest(10, 'ast', keep='first')
ast_df_temp.sort_values(by='ast', ascending=False).head(10)
```

### 4.2 Jogadores com mais rebotes na temporada

```python
reb_df_temp = temporada.groupby('player')['trb'].sum()
reb_df_temp = pd.DataFrame(reb_df_temp)
reb_df_temp
```

```python
reb_df_temp.sort_values(by='trb', ascending=False).head(10)
```

## 5. Análise por time

### 5.1 Total de pontos dos times na temporada

```python
pontos_time = temporada.groupby(['tm'])['pts'].sum()
pontos_time = pd.DataFrame(pontos_time)
pontos_time
```

### 5.2 Listar os 5 times com mais pontos

```python
top5_pontos_time = pontos_time.sort_values(by='pts', ascending=False).head(6)
top5_pontos_time
```

## 6. Coluna de Eficiência

#### *Fórmula (pts + reb + ast) / partidas

```python
plan['eff'] = (plan['pts'] + plan['trb'] + plan['ast']) / plan['g']
plan = pd.DataFrame(plan)
plan
```

```python
top5eff = plan.groupby('player')['eff'].sum()
top5eff = pd.DataFrame(top5eff)
top5eff
```

```python
top5eff = plan.sort_values('eff', ascending=False).head(5)[['player', 'eff']]
top5eff
```
